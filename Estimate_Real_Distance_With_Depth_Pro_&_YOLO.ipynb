```python
# ======================
# STEP 1: Install dependencies
# ======================
!pip install open3d numpy opencv-python pillow

# ======================
# STEP 2: Import libraries
# ======================
import numpy as np
import open3d as o3d

# ======================
# STEP 3: Load the .npz depth map safely
# ======================
depth_path = "/content/ml-depth-pro/output/IMG-20250917-WA0032.npz"

# Allow pickle because file contains object data
data = np.load(depth_path, allow_pickle=True)
print("Keys inside file:", list(data.keys()))

# Usually the depth array is stored under 'depth' or similar
# We'll check and pick the first numeric array
depth_map = None
for key in data.keys():
    arr = data[key]
    if isinstance(arr, np.ndarray) and arr.dtype != object:
        depth_map = arr
        print("✅ Depth map found in key:", key)
        break

if depth_map is None:
    raise ValueError("No valid numeric depth map found in this .npz file!")

# ======================
# STEP 4: Clean and inspect depth map
# ======================
depth_map = np.squeeze(depth_map)
H, W = depth_map.shape
print(f"Depth map shape: {H}x{W}, min={depth_map.min():.3f}, max={depth_map.max():.3f}")

# ======================
# STEP 5: Camera intrinsics (temporary)
# ======================
fx = fy = W   # approximate focal length
cx = W / 2.0
cy = H / 2.0

# ======================
# STEP 6: Convert depth map → 3D points
# ======================
points = []
for v in range(H):
    for u in range(W):
        Z = depth_map[v, u]
        if Z > 0:  # skip invalid depth
            X = (u - cx) * Z / fx
            Y = (v - cy) * Z / fy
            points.append([X, Y, Z])

points = np.array(points)
print("Total valid points:", len(points))

# ======================
# STEP 7: Create and save point cloud
# ======================
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)

output_path = "/content/ml-depth-pro/output/IMG-20250917-WA0032_pointcloud.ply"
o3d.io.write_point_cloud(output_path, pcd)
print("✅ Point cloud saved at:", output_path)
```

```python
import plotly.graph_objects as go



pcd = o3d.io.read_point_cloud("/content/ml-depth-pro/output/IMG-20250917-WA0032_pointcloud.ply")
points = np.asarray(pcd.points)

fig = go.Figure(data=[go.Scatter3d(
    x=points[:,0],
    y=points[:,1],
    z=points[:,2],
    mode='markers',
    marker=dict(
        size=1,
        color=points[:,2],   # color by depth (Z)
        colorscale='Viridis',
        opacity=0.8
    )
)])

fig.update_layout(
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    ),
    width=800,
    height=600
)

fig.show()
```
